```python
def get_match_sources(x):
    if pd.isna(x) or x.strip() == "":
        return "None"

    ids = x.split(",")

    has_s2 = any(i.startswith("S2-") for i in ids)
    has_s3 = any(i.startswith("S3-") for i in ids)

    if has_s2 and has_s3:
        return "Both"
    elif has_s2:
        return "Source2 only"
    elif has_s3:
        return "Source3 only"
    else:
        return "None"

ground_truth["match_source"] = ground_truth["matched_entity_ids"].apply(
    get_match_sources
)

print(ground_truth["match_source"].value_counts())
```

```python
from difflib import SequenceMatcher

sample_gt = ground_truth[
    ground_truth["match_count"] > 0
].sample(5000, random_state=42)

s1_names = source1.set_index("entity_id")["business_name"]
s2_names = source2.set_index("entity_id")["business_name"]
s3_names = source3.set_index("entity_id")["business_name"]

name_scores = []

for _, row in sample_gt.iterrows():

    s1_name = str(s1_names[row["source1_entity_id"]]).lower().strip()

    for matched_id in row["matched_entity_ids"].split(","):

        if matched_id.startswith("S2-"):
            matched_name = s2_names.get(matched_id, "")
        else:
            matched_name = s3_names.get(matched_id, "")

        matched_name = str(matched_name).lower().strip()

        score = SequenceMatcher(
            None,
            s1_name,
            matched_name
        ).ratio()

        name_scores.append(score)

print("Matched pairs:", len(name_scores))

pd.Series(name_scores).describe()
```

```python
import numpy as np
from difflib import SequenceMatcher

np.random.seed(42)

# Sample 5,000 Source 1 records
sample_s1 = source1.sample(
    5000,
    random_state=42
).reset_index(drop=True)

# Create country-specific Source 2 samples ONCE
s2_us = source2[source2["country"] == "US"].sample(
    5000, random_state=123
).reset_index(drop=True)

s2_india = source2[source2["country"] == "India"].sample(
    5000, random_state=456
).reset_index(drop=True)

nonmatch_name_scores = []
nonmatch_address_scores = []

for i in range(5000):

    s1 = sample_s1.iloc[i]

    if s1["country"] == "US":
        s2 = s2_us.iloc[i]
    else:
        s2 = s2_india.iloc[i]

    s1_name = str(s1["business_name"]).lower().strip()
    s2_name = str(s2["business_name"]).lower().strip()

    s1_address = str(s1["business_address"]).lower().strip()
    s2_address = str(s2["business_address"]).lower().strip()

    nonmatch_name_scores.append(
        SequenceMatcher(None, s1_name, s2_name).ratio()
    )

    nonmatch_address_scores.append(
        SequenceMatcher(None, s1_address, s2_address).ratio()
    )

print("Non-match pairs:", len(nonmatch_name_scores))

print("\nName similarity:")
print(pd.Series(nonmatch_name_scores).describe())

print("\nAddress similarity:")
print(pd.Series(nonmatch_address_scores).describe())
```

```python
import unicodedata

def normalize_unicode(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    # Remove extra spaces
    text = " ".join(text.split())

    return text.strip()
```

```python
import re

def normalize_punctuation(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Replace punctuation/symbols with spaces
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Remove extra spaces
    text = " ".join(text.split())

    return text.strip()
```

```python
for df in [source1, source2, source3]:

    df["name_normalized"] = df["name_normalized"].apply(
        normalize_punctuation
    )

    df["address_normalized"] = df["address_normalized"].apply(
        normalize_punctuation
    )
```

```python
def safe_normalize(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Normalize whitespace
    text = " ".join(text.split())

    return text.strip()
```

```python
# ============================================================
# COLAB-SAFE
# Address token overlap on only 16,690 TRUE pairs
# ============================================================

import re
import numpy as np

def address_tokens(text):
    if pd.isna(text):
        return set()

    text = str(text).lower()

    # Keep Unicode letters/numbers
    text = re.sub(r"[^\w]+", " ", text, flags=re.UNICODE)

    return set(text.split())


# S1 address lookup - only 10,000 rows
s1_address_lookup = (
    s1_test
    .set_index("entity_id")["address_normalized"]
)

# Map S1 addresses to the 16,690 true pairs
gt_s2_addr["s1_address"] = (
    gt_s2_addr["source1_entity_id"]
    .map(s1_address_lookup)
)


# S2 addresses - only 16,690 rows
s2_address_lookup = (
    true_s2_address
    .set_index("entity_id")["business_address"]
)

gt_s2_addr["s2_address"] = (
    gt_s2_addr["s2_ids"]
    .map(s2_address_lookup)
)


# Tokenize
gt_s2_addr["s1_tokens"] = (
    gt_s2_addr["s1_address"]
    .map(address_tokens)
)

gt_s2_addr["s2_tokens"] = (
    gt_s2_addr["s2_address"]
    .map(address_tokens)
)


# Jaccard similarity
def jaccard(a, b):
    if not a and not b:
        return 1.0

    if not a or not b:
        return 0.0

    return len(a & b) / len(a | b)


gt_s2_addr["address_token_similarity"] = [
    jaccard(a, b)
    for a, b in zip(
        gt_s2_addr["s1_tokens"],
        gt_s2_addr["s2_tokens"]
    )
]


print(
    gt_s2_addr[
        "address_token_similarity"
    ].describe()
)
```

```python
# ============================================================
# COLAB-SAFE
# Compare combined Name + Address scores
# ============================================================

true_scores = true_sample[
    ["name_similarity", "address_similarity"]
].copy()

true_scores["label"] = 1

false_scores = negative_pairs[
    ["name_similarity", "address_token_similarity"]
].copy()

false_scores = false_scores.rename(
    columns={
        "address_token_similarity": "address_similarity"
    }
)

false_scores["label"] = 0

score_data = pd.concat(
    [true_scores, false_scores],
    ignore_index=True
)

# Different weight combinations
score_data["score_70_30"] = (
    0.70 * score_data["name_similarity"]
    + 0.30 * score_data["address_similarity"]
)

score_data["score_50_50"] = (
    0.50 * score_data["name_similarity"]
    + 0.50 * score_data["address_similarity"]
)

score_data["score_30_70"] = (
    0.30 * score_data["name_similarity"]
    + 0.70 * score_data["address_similarity"]
)

print("========== MEAN SCORES ==========")

print("\n70% Name + 30% Address")
print(score_data.groupby("label")["score_70_30"].mean())

print("\n50% Name + 50% Address")
print(score_data.groupby("label")["score_50_50"].mean())

print("\n30% Name + 70% Address")
print(score_data.groupby("label")["score_30_70"].mean())
```

```python
# ============================================================
# COLAB-SAFE
# Threshold experiment for 30% Name + 70% Address
# ============================================================

beta = 0.5

score = score_data["score_30_70"]
actual = score_data["label"]

thresholds = [
    0.50, 0.60, 0.65, 0.70,
    0.75, 0.80, 0.85, 0.90, 0.95
]

results = []

for threshold in thresholds:

    predicted = (score >= threshold).astype(int)

    tp = ((predicted == 1) & (actual == 1)).sum()
    fp = ((predicted == 1) & (actual == 0)).sum()
    fn = ((predicted == 0) & (actual == 1)).sum()

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0

    f05 = (
        (1 + beta**2) * precision * recall
        / (beta**2 * precision + recall)
        if (precision + recall) > 0
        else 0
    )

    results.append({
        "threshold": threshold,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "precision": precision,
        "recall": recall,
        "F0.5": f05
    })

threshold_results = pd.DataFrame(results)

print(threshold_results.to_string(index=False))
```

```python
# ============================================================
# SAFE ADDRESS SIMILARITY
# Only ~6,000 rows
# ============================================================

import re

def address_tokens(text):
    if pd.isna(text) or text == "":
        return set()

    text = str(text).lower()

    # Keep Unicode letters/numbers
    text = re.sub(r"[^\w]+", " ", text, flags=re.UNICODE)

    return set(text.split())


def jaccard_similarity(a, b):
    if not a and not b:
        return 1.0

    if not a or not b:
        return 0.0

    return len(a & b) / len(a | b)
```
