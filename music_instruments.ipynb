```python
!pip install transformers
from transformers import pipeline
generator = pipeline('text-generation', model='gpt2')
qanda = pipeline('question-answering')
summarizer = pipeline('summarization')
unmasker = pipeline('fill-mask')
image_describe = pipeline('image-to-text')
vqa = pipeline('vqa')
object_detection = pipeline('object-detection')
image_classify = pipeline('image-classification')
```

```python
text = "Once a time there lived a business, pire boss call"
generator(text, max_length=50, num_return_sequences=1)
```

```python
music = pipeline('text-to-audio', model='facebook/musicgen-small')
```

```python
prompt = " A dreamy, slightly melancholic lo-fi hip-hop track with a gentle, head-nodding rhythm. The main instrumentation should feature a warm, heavily compressed vinyl-style drum loop (with a subtle crackle), a deep, resonant sub-bass line, and a sampled electric piano (Rhodes) playing a simple, soulful chord progression. The melody should be carried by a hazy, reverbed analog synthesizer pad that slowly evolves. The tempo should be around 85 BPM. Imagine rain hitting a window late at night"
op=music(prompt)
Audio(op['audio'].flatten(),rate = op['sampling_rate'],autoplay=True)
```
